```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
import numpy as np
```

```python


# fetch dataset
bank_marketing = fetch_ucirepo(id=222)

# data (as pandas dataframes)
X = bank_marketing.data.features
y = bank_marketing.data.targets

# metadata
print(bank_marketing.metadata)

# variable information
print(bank_marketing.variables)
```

```python
X
```

```python
y
```

```python
df = X.join(y)
df['subscription'] = df['y'].map({'yes': 1, 'no': 0})
df.drop(columns=['y'], inplace=True)
df.head()
```

## Tratamento de NaNs


```python
df.isna().sum()
```

```python
df['contact'].replace(np.nan, 'unknown', inplace=True)
df['job'].replace(np.nan, 'unknown', inplace=True)
df['education'].replace(np.nan, 'unknown', inplace=True)
df['poutcome'].replace(np.nan, 'unknown', inplace=True)
```

## Altera nomes

```python
df.rename(columns={'default': 'credit_default',
                   'housing': 'housing_loan',
                   'loan': 'personal_loan',
                   'day_of_week': 'last_contact_day',
                   'month': 'last_contact_month',
                   'duration': 'last_contact_duration'},
          inplace=True)
```

```python
df.to_csv('bank_marketing.csv', index=False)
```
